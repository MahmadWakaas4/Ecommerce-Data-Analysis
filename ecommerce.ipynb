```python
# Importing Libraries

import pandas as pd
import numpy as np 
```

```python
# Laod the data

df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# Explore the dataset

df.head()
```

```python
df.info()
```

```python
# Statistical Summary

df.describe()
```

```python
# Data Shape

df.shape
```

```python
# Missing Values
df.isnull().sum()
```

```python
# Duplicate Records
df.duplicated().sum()
```

```python
# Remove duplicates

df.drop_duplicates(inplace=True)
```

```python
# Replace Invalid 
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
# Remove EXtra Spaces

df=df.apply(lambda x:x.str.strip()if x.dtype=='object' else x)
```

```python
# Standardize Text Format

df['Customer_Name']=df['Customer_Name'].str.title()
df['City']=df['City'].str.title()
df['State']=df['State'].str.title()
```

```python
# Validate Email ID

df=df[df['Email'].str.contains('@',na=False)]
```

```python
# Convert Date Columns

df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# Convert Numeric Values

cols=['Qty','Unit_Price','Discount']
for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
# Remove Invalid Quantity

df=df[df['Qty']>0]
```

```python
# Fill Missing Values

df['Discount'].fillna(0)
df['Phone'].fillna('Unknown')
df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Feature Engineering
# 1.Total Sales
df['Sales']=df['Qty']*df['Unit_Price']
```

```python
# 2.Net Amount

df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
# 3.Profit (Example: 20%)

df['Profit']=df['Net_Amount']*0.20
```

```python
# 4.Order Month

df['Month']=df['Order_Date'].dt.month_name()
```

```python
# 5.Order Year
df['Year']=df['Order_Date'].dt.year
```

```python
# 6. Weekday

df['Weekday']=df['Order_Date'].dt.day_name()
```

```python
# Final Data Validation

df.info()
```

```python
# Export Clean Data

df.to_csv('Clean_Ecommerce.csv',index=False)
```
